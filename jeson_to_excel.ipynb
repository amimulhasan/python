```python
import pandas as pd
import json

# JSON file load
with open('/content/data_info (6).json', 'r') as f:
    data = json.load(f)

# DataFrame convert
df = pd.DataFrame(data)

# Excel file save
df.to_excel('output.xlsx', index=False)

print("JSON successfully converted to Excel")
```
